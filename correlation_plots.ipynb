# Plotting correlations from minors to majors

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import scipy
from sklearn.metrics import r2_score
```

```python
# Read data
df = pd.read_csv("data/minors_to_majors.csv")
df
```

## Plots against MLB career wOBA

```python
x_cols = [
    "Age",
    "Season",
    "minor_PA",
    "minor_BB%",
    "minor_K%",
    "minor_BB/K",
    "minor_BA",
    "minor_OBP",
    "minor_SLG",
    "minor_OPS",
    "minor_ISO",
    "minor_Speed",
    "minor_BABIP",
    "minor_wSB",
    "minor_wRAA",
    "minor_wOBA",
    "minor_wRC",
    "minor_wRC+",
]
y_col = "major_wOBA"

fig, axes = plt.subplots(9, 2, figsize=(20, 30), sharey=True)
axes = axes.ravel()

for ax, x in zip(axes, x_cols):
    sns.regplot(data=df, x=x, y=y_col, ax=ax)
    pearson_r = scipy.stats.pearsonr(x=df[x], y=df[y_col])[0]
    r = r2_score(y_true=df[x], y_pred=df[y_col])
    ax.text(
        0.05,
        0.95,
        f"$R^2$ = {r:.2f}\n r = {pearson_r:.2f}\n ",
        transform=ax.transAxes,
        ha="left",
        va="top",
        fontsize=12,
        bbox=dict(boxstyle="round,pad=0.5", facecolor="white", alpha=0.7),
    )
    ax.set_title(x)

[fig.delaxes(ax) for ax in axes.flatten() if not ax.has_data()]  # Remove unusued plots

plt.tight_layout()
plt.show()
```

```python
# Distribution by Team
order = sorted(df["minor_Team"].unique())

plt.figure(figsize=(12, 5))
sns.boxplot(data=df, x="minor_Team", y="major_wOBA", order=order)
plt.title("Major League wOBA Distributed by Minor League Teams")
plt.tight_layout()
plt.show()
```
